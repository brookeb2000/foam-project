```python
# The purpose of this model is to import the GPR model from
# the models folder as well as the scaler and test that it works
```

```python
import pickle

# Load the scaler
scaler = pickle.load(open('../models/scaler_wet_strength.pkl', 'rb'))

# Load the model
model = pickle.load(open('../models/gpr_wet_strength_model.pkl', 'rb'))
```

```python
# scale a new imput of interest

import pandas as pd
import numpy as np

formula_cols = ["TEM", "Joncryl", "NC513", "PEG-DE", "NGDE"]

# To scale NEW data, use transform():
new_input = pd.DataFrame([[8, np.nan, np.nan, np.nan, np.nan]], columns=formula_cols)
new_scaled = scaler.transform(new_input)

# Convert to DataFrame if you want column names:
new_scaled_df = pd.DataFrame(new_scaled, columns=formula_cols)
print(new_scaled_df)
```

```python
# make inference with the input of interest

mean, std = model.predict([[.5, 0, 0, 0, 0]], return_std=True)
print(f"Mean: {mean[0]:.6f}, Std: {std[0]:.6f}")
```
